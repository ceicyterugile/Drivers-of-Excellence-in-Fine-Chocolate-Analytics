## **A/B Testing**: Vanilla vs. Non-Vanilla Chocolate Bars

### Subtask:
Perform an independent samples t-test to compare the average ratings of chocolate bars with and without vanilla and determine if the difference is statistically significant.

**Reasoning**:
To perform the A/B test, I need to:
1.  Separate the DataFrame into two groups based on the presence of vanilla in the 'ingredients' column (similar to the previous SQL query).
2.  Extract the 'rating' data for each group.
3.  Use `scipy.stats.ttest_ind` to conduct an independent samples t-test.
4.  Interpret the resulting t-statistic and p-value to determine statistical significance.

```python
import pandas as pd
import scipy.stats as stats

# Load the dataframe (ensure df is always available)
df = pd.read_csv('/content/chocolate_bars.csv')

# Re-apply cleaning for consistency with earlier steps
median_num_ingredients = df['num_ingredients'].median()
df['num_ingredients'] = df['num_ingredients'].fillna(median_num_ingredients)
df['ingredients'] = df['ingredients'].fillna('Unknown') # Fill actual NaNs

# Ensure 'ingredients' column is string type for .str.contains()
df['ingredients'] = df['ingredients'].astype(str)

# Group data based on presence of vanilla
vanilla_group = df[df['ingredients'].str.contains('v', case=False)]['rating'].dropna()
non_vanilla_group = df[~df['ingredients'].str.contains('v', case=False)]['rating'].dropna()

# Perform independent samples t-test
t_statistic, p_value = stats.ttest_ind(vanilla_group, non_vanilla_group, equal_var=False) # Assuming unequal variances

print(f"T-statistic: {t_statistic:.3f}")
print(f"P-value: {p_value:.3f}")

# Interpret the results
alpha = 0.05
if p_value < alpha:
    print(f"\nWith a p-value of {p_value:.3f} (less than {alpha}), we reject the null hypothesis.")
    print("There is a statistically significant difference in average ratings between chocolate bars with and without vanilla.")
    if vanilla_group.mean() > non_vanilla_group.mean():
        print(f"Chocolate bars with vanilla have a significantly higher average rating ({vanilla_group.mean():.3f}).")
    else:
        print(f"Chocolate bars without vanilla have a significantly higher average rating ({non_vanilla_group.mean():.3f}).")
else:
    print(f"\nWith a p-value of {p_value:.3f} (greater than or equal to {alpha}), we fail to reject the null hypothesis.")
    print("There is no statistically significant difference in average ratings between chocolate bars with and without vanilla.")
```

## Extract and Count Individual Ingredients

### Subtask:
Parse the 'ingredients' column to extract individual ingredients and count their occurrences across all chocolate bars.


**Reasoning**:
The first step is to fill any missing values in the 'ingredients' column with an empty string, as specified in the instructions, to prevent errors during string processing.



```python
df['ingredients'] = df['ingredients'].fillna('')
print("Missing values in 'ingredients' after filling: ", df['ingredients'].isnull().sum())
```

**Reasoning**:
To prepare the 'ingredients' for analysis, I will define a function to clean and extract individual ingredients from the 'ingredients' string, then apply this function to create a new 'parsed_ingredients' column.



```python
import re

def clean_and_extract_ingredients(ingredients_str):
    if not isinstance(ingredients_str, str):
        return []
    # Convert to lowercase
    ingredients_str = ingredients_str.lower()
    # Remove numeric characters (e.g., '1.', '2.')
    ingredients_str = re.sub(r'\d+\.?\s*', '', ingredients_str)
    # Replace common separators with a single comma
    ingredients_str = re.sub(r'[,;/]| and ', ',', ingredients_str)
    # Split by comma, strip whitespace, and remove empty strings
    ingredients = [ing.strip() for ing in ingredients_str.split(',') if ing.strip()]
    return ingredients

df['parsed_ingredients'] = df['ingredients'].apply(clean_and_extract_ingredients)
print("First 5 rows of 'parsed_ingredients' column:")
print(df['parsed_ingredients'].head())
```

```python
import pandasql as ps

# Create a temporary DataFrame without the 'parsed_ingredients' column for pandasql
df_temp = df.drop(columns=['parsed_ingredients'], errors='ignore')

sql_query_origin_trends = """
SELECT
    bean_origin,
    year_reviewed,
    COUNT(id) AS Total_Reviews_in_Year,
    ROUND(AVG(rating), 3) AS Average_Rating_in_Year
FROM
    df_temp
WHERE
    bean_origin IS NOT NULL
    AND year_reviewed IS NOT NULL
    AND rating IS NOT NULL
GROUP BY
    bean_origin,
    year_reviewed
HAVING
    COUNT(id) >= 3
    AND bean_origin IN (
        SELECT bean_origin
        FROM df_temp
        GROUP BY bean_origin
        HAVING COUNT(id) >= 5
        ORDER BY AVG(rating) DESC
        LIMIT 10
    )
ORDER BY
    bean_origin ASC,
    year_reviewed ASC;
"""

origin_trends_filtered_sql = ps.sqldf(sql_query_origin_trends, globals())
print("Bean Origin Trends Over Time (Filtered for >= 3 reviews and Top 10 origins with >= 5 overall reviews):")
print(origin_trends_filtered_sql)

print("\n--- SQL Query Explanation ---")
print("This SQL query aims to analyze the evolution of chocolate bar ratings and review counts for the top 10 most highly-rated bean origins over time. Here's a breakdown:")
print("\n1.  **SELECT Clause**: Specifies the columns to be retrieved:")
print("    *   `bean_origin`: The origin of the cocoa beans.")
print("    *   `year_reviewed`: The year the chocolate bar was reviewed.")
print("    *   `COUNT(id) AS Total_Reviews_in_Year`: Counts the number of reviews for each bean origin in a given year.")
print("    *   `ROUND(AVG(rating), 3) AS Average_Rating_in_Year`: Calculates the average rating for each bean origin in a given year, rounded to 3 decimal places.")

print("\n2.  **FROM Clause**: Indicates the table (`df_temp`) from which data is being retrieved.")

print("\n3.  **WHERE Clause**: Filters the rows based on certain conditions:")
print("    *   `bean_origin IS NOT NULL`, `year_reviewed IS NOT NULL`, `rating IS NOT NULL`: Ensures that only records with complete information for these key columns are considered.")

print("\n4.  **GROUP BY Clause**: Groups the results by `bean_origin` and `year_reviewed`. This means the aggregate functions (COUNT and AVG) will operate independently for each unique combination of origin and year.")

print("\n5.  **HAVING Clause**: Filters the *grouped* results based on conditions:")
print("    *   `COUNT(id) >= 3`: Only includes groups (origin-year combinations) that have at least 3 reviews, ensuring a minimum level of data stability and reliability for the average rating.")
print("    *   `bean_origin IN (...)`: This is a subquery that further filters the results to include only the `bean_origin` values that are among the overall top 10 highest-rated origins. The subquery itself:")
print("        *   Selects `bean_origin` from `df_temp`.")
print("        *   Groups them by `bean_origin`.")
print("        *   Requires `COUNT(id) >= 5` for each origin in the subquery to ensure reliability for the 'top 10' calculation. This is the change from the previous query.")
print("        *   Orders these by `AVG(rating) DESC` to find the highest-rated.")
print("        *   `LIMIT 10` picks only the top 10 origins overall.")

print("\n6.  **ORDER BY Clause**: Sorts the final results first by `bean_origin` in ascending order, and then by `year_reviewed` in ascending order within each origin, making the trends easier to follow chronologically.")

print("\nIn summary, this query provides a clean, time-series view of the average ratings and review volumes for the most consistently high-performing cocoa bean origins, allowing us to identify evolutionary trends in their quality and popularity.")
```

## Interpreting the T-statistic and P-value

From the independent samples t-test conducted previously, we obtained the following results:

*   **T-statistic: -6.735**
*   **P-value: 0.000**
*   **Significance Level (alpha): 0.05**

### Explanation of the T-statistic

The T-statistic measures the size of the difference relative to the variation in your sample data. In this context:

*   A **T-statistic of -6.735** indicates a substantial difference in the average ratings between chocolate bars with vanilla and those without. The negative sign implies that the average rating of the `vanilla_group` (which came first in the `ttest_ind` function call) is lower than the average rating of the `non_vanilla_group`.
*   The magnitude (absolute value) of 6.735 is relatively large, suggesting that the observed difference is unlikely to be due to random chance alone, especially given the sample sizes.

### Explanation of the P-value

The P-value is the probability of observing a test statistic as extreme as, or more extreme than, the one calculated from the sample data, assuming that the null hypothesis is true. In this context:

*   The **P-value of 0.000** means there is an extremely small probability (effectively 0%) of observing a difference in average ratings of this magnitude (or greater) if there were truly no difference between chocolate bars with and without vanilla in the overall population.
*   A very small P-value (typically less than the chosen alpha level) suggests that the observed data is inconsistent with the null hypothesis.

### Statistical Decision and Implication

*   **Null Hypothesis (H0):** There is no difference in the average ratings between chocolate bars with and without vanilla.
*   **Alternative Hypothesis (Ha):** There is a statistically significant difference in the average ratings between chocolate bars with and without vanilla.

Since our **P-value (0.000) is less than the chosen significance level (alpha = 0.05)**, we **reject the null hypothesis**.

This statistical decision implies that there is a **statistically significant difference in average ratings between chocolate bars with and without vanilla**. Specifically, as indicated by the negative t-statistic and the mean comparison in the previous output, **chocolate bars without vanilla have a significantly higher average rating (3.223) compared to those with vanilla (3.033)**. This suggests that, from a statistical standpoint, the presence of vanilla is associated with lower average ratings in chocolate bars.
